# Ejercicio 2. Transcripción del ADN a ARN

### 1. Usa la siguiente secuencia de ADN: 5’ – ATG CCT GAA TGC – 3’ / 3’ – TAC GGA CTT ACG – 5’.

### 2. Identifica cuál es la cadena molde.

La cadena molde es  **3’ – TAC GGA CTT ACG – 5’**, debido a que va de 3' a 5', además observamos que la otra hebra es la codificante, ya que empieza por ***ATG*** y va de 5' a 3'.

### 3. Obtén el transcrito de ARN correspondiente (recuerda que el ARN se sintetiza en dirección 5’ → 3’ y utiliza uracilo en lugar de timina).

El transcrito resultante de la hebra molde sería lo mismo que la hebra codificante pero cambiando la Timina por Uracilo: **5’ – AUG CCU GAA UGC – 3’**

### 4. Explica cuál sería la región promotora y cuál la región codificante.

En este caso la secuencia ya empieza con el codón *ATG* por lo que vemos únicamente la región promotora.

### Extensión con Biopython:

```python
from pathlib import Path
from Bio import SeqIO

fasta = Path("../Ficheros-20260927/Fasta/lacZ_Ecoli_NC_000913.3.fasta")
record = SeqIO.read(fasta, "fasta")

# La secuencia del FASTA está en orientación codificante 5' -> 3'
adn_codificante = record.seq

# ARNm obtenido directamente de la cadena codificante
arnm = adn_codificante.transcribe()

# Cadena molde complementaria, orientada 3' -> 5'
cadena_molde = adn_codificante.complement()

# ARNm correcto a partir de la cadena molde
arnm_desde_molde = cadena_molde.complement().transcribe()

# Resultado incorrecto si no se tiene en cuenta la orientación
arnm_mal_orientado = cadena_molde.transcribe()

print("Archivo leído:", fasta)
print("ARNm correcto:", arnm)
print("Cadena molde:", cadena_molde)
print("ARNm desde la cadena molde:", arnm_desde_molde)
print("ARNm mal orientado:", arnm_mal_orientado)
```

La orientación es fundamental: al usar la hebra correcta se obtiene el ARNm esperado, mientras que al transcribir la hebra con orientación incorrecta aparece una secuencia diferente y no válida para la traducción.
