# Ejercicio 4. Splicing alternativo

**Objetivo:** comprender cómo un mismo gen puede generar varias proteínas.

## Instrucciones

### 1. Considera un gen con 5 exones: Exón 1 – Exón 2 – Exón 3 – Exón 4 – Exón 5.

### 2. Diseña al menos dos combinaciones de splicing alternativo (por ejemplo, 1-2-4-5 o 1-3-5).

1-2-3-4-5 y 2-4-5

### 3. Explica qué diferencias esperarías en las proteínas resultantes.

La primera tiene todos los exónes por lo que resultaría en una proteína completa, la otra sin embargo al saltarse los exónes 1 y 2 resultaría en una proteína más corta sin los segmentos de aminoácidos correspondientes a esos exónes.

### 4. Reflexiona: ¿por qué este mecanismo aumenta la diversidad proteica sin necesidad de más genes?

Debido a que a partir de un mismo gen se pueden extraer distintos ARNm al realizar distintas combinaciones entre los exónes.

## Extensión con Biopython / bases de datos:

```python
from urllib.request import Request, urlopen
import json

# FGFR2 en Ensembl
gene_id = "ENSG00000066468"
url = f"https://rest.ensembl.org/lookup/id/{gene_id}?expand=1"

request = Request(url, headers={"Content-Type": "application/json"})

with urlopen(request) as respuesta:
    gen = json.load(respuesta)

print("Gen:", gen["display_name"])
print("Número de transcritos:", len(gen["Transcript"]))

for transcrito in gen["Transcript"]:
    exones = transcrito.get("Exon", [])
    traduccion = transcrito.get("Translation")

    longitud = sum(
        abs(exon["end"] - exon["start"]) + 1
        for exon in exones
    )

    if traduccion:
        proteina = f"{traduccion['length']} aminoácidos"
    else:
        proteina = "sin proteína"

    print(
        transcrito["id"],
        "- exones:", len(exones),
        "- longitud:", longitud, "nt",
        "- proteína:", proteina
    )
```

Los transcritos pueden tener diferente número de exones y distinta longitud. Esto puede producir proteínas con diferentes dominios, tamaños y funciones. Algunos transcritos pueden no producir ninguna proteína.
