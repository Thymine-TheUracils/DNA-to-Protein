### 1. Escoge una secuencia de ADN (codificada en un fichero fasta de algún banco de datos o página especializada públicos).

### 2. Paso 1: replica la secuencia indicando las dos nuevas hebras.



### 3. Paso 2: transcribe la cadena molde en ARNm.



### 4. Paso 3: traduce el ARNm en una cadena de aminoácidos.



### 5. Reflexiona: ¿qué punto del proceso es más vulnerable a errores que afecten a la función de la proteína?



### 6. Programa un pipeline que realice los tres procesos (replicación, transcripción, y traducción.



### Extensión con Biopython:

```python
from Bio import SeqIO


def pipeline(fasta):
    print("[INICIO] Leyendo el archivo FASTA...")
    record = SeqIO.read(fasta, "fasta")
    codificante = record.seq

    print(f"[INFO] Secuencia cargada: {record.id}")
    print(f"[INFO] Longitud: {len(codificante)} nucleótidos")

    print("\n[PASO 1] Replicación")
    molde = codificante.complement()
    nueva_hebra_1 = molde
    nueva_hebra_2 = codificante

    print("[INFO] Nueva hebra 1:")
    print(nueva_hebra_1)
    print("[INFO] Nueva hebra 2:")
    print(nueva_hebra_2)

    print("\n[PASO 2] Transcripción")
    arnm = molde.complement().transcribe()
    print("[INFO] ARNm obtenido:")
    print(arnm)

    print("\n[PASO 3] Traducción")
    proteina = arnm.translate(to_stop=True)
    print("[INFO] Proteína obtenida:")
    print(proteina)

    print("\n[FIN] Pipeline completado.")

    return nueva_hebra_1, nueva_hebra_2, arnm, proteina


fasta = "../Ficheros-20260927/Fasta/lacZ_Ecoli_NC_000913.3.fasta"

pipeline(fasta)
```
