# Ejercicio 3. Traducción del ARNm a proteína

**Objetivo:** aplicar el código genético y reflexionar sobre mutaciones.

## Instrucciones

### 1. Usa el siguiente transcrito de ARN

```text
5′-AUG UAU GCU UAA-3′
```

Sin espacios: `AUGUAUGCUUAA`. Se lee en dirección 5′→3′, en grupos de tres bases llamados codones.

### 2. Identifica el codón de inicio y el codón de paro

El codón de inicio es **AUG** y el codón de paro es **UAA**.

### 3. Traduce la secuencia en una cadena de aminoácidos

| Codón | Resultado |
| --- | --- |
| AUG | Metionina (Met, M) |
| UAU | Tirosina (Tyr, Y) |
| GCU | Alanina (Ala, A) |
| UAA | Paro |

La cadena es **Met–Tyr–Ala**, o `MYA` en el código de una letra. El codón de paro no añade un aminoácido.

### 4. Reflexiona: ¿qué pasaría si el codón de inicio mutara de AUG a GUG? ¿Qué ocurriría si el codón de paro desapareciera por mutación?

**AUG a GUG:** AUG es la señal habitual para empezar la traducción. Si cambia a GUG, el ribosoma podría no reconocer dónde empezar y la proteína podría no producirse.

**Pérdida del codón de paro:** el ribosoma perdería la señal para detenerse y podría seguir añadiendo aminoácidos hasta encontrar otro codón de paro. La proteína sería más larga de lo esperado y podría no funcionar correctamente.

## Extensión con Biopython

Utiliza el módulo Bio.Seq para traducir automáticamente el ARNm a una secuencia proteica y comprueba si el resultado coincide con el tuyo.

````python
from Bio.Seq import Seq

arnm = "AUGUAUGCUUAA"
proteina = str(Seq(arnm).translate(to_stop=True))

print(f"ARNm: 5′-{arnm}-3′")
print(f"Proteína: {proteina} (Met–Tyr–Ala)")
````

Coincide con el resultado manual: MYA
