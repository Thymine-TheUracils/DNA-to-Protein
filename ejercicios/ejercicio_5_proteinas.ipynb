# Ejercicio 5. Introducción a las proteínas

**Objetivo:** relacionar secuencia, estructura y función.

## Instrucciones

### 1. Considera la siguiente secuencia de aminoácidos

```text
Met–Ile–Ser–Gly–Val–Lys–His
```

En el código de una letra: `MISGVKH`.

### 2. Identifica el extremo N y el extremo C de la cadena

El **extremo N** corresponde a la metionina (Met), con el grupo amino libre. El **extremo C** corresponde a la histidina (His), con el grupo carboxilo libre.

```text
N → Met–Ile–Ser–Gly–Val–Lys–His → C
```

### 3. Reflexiona: ¿cómo influye el orden de los aminoácidos en la estructura final de la proteína? ¿Qué ocurriría si hubiera una mutación que cambiara un aminoácido hidrofóbico por uno hidrofílico en una región interna de la proteína?

**Orden de los aminoácidos:** determina qué grupos pueden interactuar durante el plegamiento. Por eso influye en la forma final y en la función de la proteína.

**Cambio en una región interna:** en una proteína soluble, sustituir un aminoácido hidrofóbico del interior por uno hidrofílico podría desestabilizar el núcleo y alterar el plegamiento o la función. El efecto depende de la posición y de las interacciones que pueda formar el nuevo aminoácido.

## Extensión con bioinformática

Busca en el Protein Data Bank (PDB) la estructura de una proteína conocida y observa cómo los aminoácidos se organizan en hélices alfa y láminas beta. Reflexiona sobre cómo una mutación puntual podría afectar al plegamiento.

**Proteína elegida:** ubiquitina, entrada [1UBQ del PDB](https://www.rcsb.org/structure/1UBQ). Su estructura incluye una hélice alfa y una lámina beta formada por cinco hebras. Las hélices tienen forma de espiral; las hebras beta se disponen unas junto a otras formando una lámina. Puedes verlas en el [visor 3D de 1UBQ](https://www.rcsb.org/3d-view/1UBQ).

**Reflexión:** una mutación puntual que sustituya un aminoácido del núcleo hidrofóbico por uno hidrofílico podría alterar las interacciones que mantienen juntas estas estructuras y reducir la estabilidad de la proteína. Es una posibilidad; para conocer el efecto concreto habría que estudiar la posición y la sustitución.
